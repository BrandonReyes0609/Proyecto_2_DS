````python
import re
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")


TRAIN_PATH = "./sample_data/trainlimpio.csv"
TEST_PATH  = "./sample_data/testlimpio.csv"
DATE_COL   = "date_id"

instruments = {
    'JPX_Gold_Standard_Futures': ['Close', 'High', 'Low', 'Open'],
    'JPX_Gold_Mini_Futures': ['settlement_price', 'High', 'Low', 'Close', 'Open'],
}

MULTI_TARGET = False

LOOKBACK   = 64
BATCH_SIZE = 32
EPOCHS     = 200
LR         = 2e-3
HIDDEN     = 512
LAYERS     = 2
PATIENCE   = 100
CONST_THR  = 1e-8
DEVICE     = "cuda" if torch.cuda.is_available() else "cpu"
SEED       = 42

np.random.seed(SEED)
torch.manual_seed(SEED)


def plot_loss_curve(loss_hist, title):
    plt.figure(figsize=(8,4))
    plt.plot(loss_hist, lw=2)
    plt.xlabel("Época")
    plt.ylabel("MSE (train)")
    plt.title(title)
    plt.tight_layout()
    plt.show()
    plt.close()

def plot_series(dates, y_true, y_pred, title):
    plt.figure(figsize=(12,4))
    if dates is not None:
        plt.plot(dates, y_true, label="Real", lw=1.5)
        plt.plot(dates, y_pred, label="Predicho", lw=1.5)
        plt.xticks(rotation=45)
    else:
        plt.plot(y_true, label="Real", lw=1.5)
        plt.plot(y_pred, label="Predicho", lw=1.5)
    plt.title(title)
    plt.legend()
    plt.tight_layout()
    plt.show()
    plt.close()

def plot_scatter(y_true, y_pred, title):
    plt.figure(figsize=(5,5))
    plt.scatter(y_true, y_pred, s=10, alpha=0.7)
    lims = [
        np.min([plt.xlim()[0], plt.ylim()[0]]),
        np.max([plt.xlim()[1], plt.ylim()[1]])
    ]
    plt.plot(lims, lims, 'k--', lw=1)
    plt.xlim(lims); plt.ylim(lims)
    plt.xlabel("Real"); plt.ylabel("Predicho")
    plt.title(title)
    plt.tight_layout()
    plt.show()
    plt.close()

def plot_residuals(dates, residuals, title_prefix):
    plt.figure(figsize=(12,3))
    if dates is not None:
        plt.plot(dates, residuals, lw=1)
        plt.xticks(rotation=45)
    else:
        plt.plot(residuals, lw=1)
    plt.axhline(0, color='k', lw=1)
    plt.title(f"{title_prefix} - Residuales (línea)")
    plt.tight_layout()
    plt.show()
    plt.close()

    plt.figure(figsize=(6,4))
    plt.hist(residuals, bins=30, alpha=0.85)
    plt.title(f"{title_prefix} - Residuales (histograma)")
    plt.tight_layout()
    plt.show()
    plt.close()


def make_sequences(X, y, lookback):
    Xs, ys = [], []
    for i in range(len(X) - lookback):
        Xs.append(X[i:i+lookback, :])
        ys.append(y[i+lookback, :])
    return np.array(Xs, dtype=np.float32), np.array(ys, dtype=np.float32)

def drop_all_nan(train, test, date_col):
    all_nan_cols = [c for c in train.columns if c != date_col and train[c].isna().all()]
    all_nan_test = [c for c in test.columns if c != date_col and test[c].isna().all()]
    all_nan_cols = sorted(set(all_nan_cols + all_nan_test))
    if all_nan_cols:
        train = train.drop(columns=all_nan_cols, errors="ignore")
        test  = test.drop(columns=all_nan_cols, errors="ignore")
    return train, test

def drop_near_constant(train, test, date_col, thr=CONST_THR):
    stds = train.drop(columns=[date_col]).std(axis=0)
    near_const = stds[stds <= thr].index.tolist()
    keep = [c for c in train.columns if c not in near_const]
    return train[keep], test[keep]

def build_target_columns(instruments_dict):
    cols = []
    for base, fields in instruments_dict.items():
        for f in fields:
            cols.append(f"{base}_{f}")
    seen = set(); ordered = []
    for c in cols:
        if c not in seen:
            seen.add(c); ordered.append(c)
    return ordered

class GRUForecaster(nn.Module):
    def __init__(self, input_size, out_dim=1, hidden=128, layers=1):
        super().__init__()
        self.gru = nn.GRU(input_size=input_size, hidden_size=hidden, num_layers=layers, batch_first=True)
        self.fc  = nn.Linear(hidden, out_dim)
        self.out_act = nn.Sigmoid()
    def forward(self, x):
        out, _ = self.gru(x)
        out = out[:, -1, :]
        out = self.fc(out)
        return self.out_act(out)

def train_eval_model(X_train, y_train, X_test, y_test, out_dim=1):
    Xtr_t = torch.tensor(X_train, dtype=torch.float32)
    ytr_t = torch.tensor(y_train, dtype=torch.float32)

    train_loader = DataLoader(TensorDataset(Xtr_t, ytr_t), batch_size=BATCH_SIZE, shuffle=True)

    model = GRUForecaster(input_size=X_train.shape[-1], out_dim=out_dim, hidden=HIDDEN, layers=LAYERS).to(DEVICE)
    crit   = nn.MSELoss()
    opt    = torch.optim.Adam(model.parameters(), lr=LR)

    best_loss = float("inf"); wait = 0; best_state = None
    loss_hist = []
    for ep in range(1, EPOCHS+1):
        model.train(); epoch_loss = 0.0
        for xb, yb in train_loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            opt.zero_grad()
            pred = model(xb)
            loss = crit(pred, yb)
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            opt.step()
            epoch_loss += loss.item() * xb.size(0)
        epoch_loss /= len(train_loader.dataset)
        loss_hist.append(epoch_loss)

        if epoch_loss < best_loss - 1e-6:
            best_loss = epoch_loss; wait = 0
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
        else:
            wait += 1

        if ep == 1 or ep % 5 == 0 or ep == EPOCHS:
            print(f"Época {ep:03d}/{EPOCHS}  Train MSE: {epoch_loss:.6f}  (best {best_loss:.6f})")
        if wait >= PATIENCE:
            print(f"Early stopping en época {ep}."); break

    if best_state is not None:
        model.load_state_dict({k: v.to(DEVICE) for k, v in best_state.items()})

    model.eval()
    with torch.no_grad():
        preds = []
        for i in range(0, len(X_test), 1024):
            xb = torch.tensor(X_test[i:i+1024], dtype=torch.float32).to(DEVICE)
            preds.append(model(xb).cpu().numpy())
        yhat = np.vstack(preds)
    return model, yhat, loss_hist

train_df = pd.read_csv(TRAIN_PATH)
test_df  = pd.read_csv(TEST_PATH)
train_df = train_df.sort_values(DATE_COL).reset_index(drop=True)
test_df  = test_df.sort_values(DATE_COL).reset_index(drop=True)

extra_in_test = [c for c in test_df.columns if c not in train_df.columns]
if extra_in_test:
    print("Eliminando columnas extra en test:", extra_in_test)
    test_df = test_df.drop(columns=extra_in_test, errors="ignore")

common_cols = [c for c in train_df.columns if c in test_df.columns]
train_df = train_df[common_cols]
test_df  = test_df[common_cols]
assert list(train_df.columns) == list(test_df.columns), "Train/Test no alinean columnas"

train_df = train_df.fillna(method="ffill").fillna(method="bfill")
test_df  = test_df.fillna(method="ffill").fillna(method="bfill")

for c in train_df.columns:
    if train_df[c].dtype == "bool":
        train_df[c] = train_df[c].astype(int)
for c in test_df.columns:
    if test_df[c].dtype == "bool":
        test_df[c] = test_df[c].astype(int)

train_df, test_df = drop_all_nan(train_df, test_df, DATE_COL)
train_df, test_df = drop_near_constant(train_df, test_df, DATE_COL, CONST_THR)


ALL_TARGETS = build_target_columns(instruments)
ALL_TARGETS = [t for t in ALL_TARGETS if t in train_df.columns and t in test_df.columns]
assert len(ALL_TARGETS) > 0, "No se encontraron columnas objetivo en los CSV según 'instruments'."

print(f"Targets detectados ({len(ALL_TARGETS)}):")
print(ALL_TARGETS)

num_cols = [c for c in train_df.columns if c != DATE_COL and np.issubdtype(train_df[c].dtype, np.number)]
feature_cols = [c for c in num_cols if (c not in ALL_TARGETS) ] if MULTI_TARGET else num_cols

x_scaler = MinMaxScaler()
X_train_all = x_scaler.fit_transform(train_df[feature_cols].values)
X_test_all  = x_scaler.transform(test_df[feature_cols].values)

test_dates = None
if DATE_COL in test_df.columns:
    test_dates = test_df[DATE_COL].iloc[LOOKBACK:].to_numpy()


if MULTI_TARGET:
    print("\n=== MODO MULTI-TARGET ===")
    y_scaler_global = MinMaxScaler()
    y_train_all = y_scaler_global.fit_transform(train_df[ALL_TARGETS].values)
    y_test_all  = y_scaler_global.transform(test_df[ALL_TARGETS].values)

    Xtr, ytr = make_sequences(X_train_all, y_train_all, LOOKBACK)
    Xte, yte = make_sequences(X_test_all,  y_test_all,  LOOKBACK)
    print("Shapes -> Xtr:", Xtr.shape, "ytr:", ytr.shape, "| Xte:", Xte.shape, "yte:", yte.shape)

    model, yhat_scaled, loss_hist = train_eval_model(Xtr, ytr, Xte, yte, out_dim=len(ALL_TARGETS))
    yhat_inv = y_scaler_global.inverse_transform(yhat_scaled)
    yte_inv  = y_scaler_global.inverse_transform(yte)

    plot_loss_curve(loss_hist, "Pérdida (Train MSE) - Multi-target")

    maes = []; rmses = []
    for j, tname in enumerate(ALL_TARGETS):
        y_true = yte_inv[:, j]
        y_pred = yhat_inv[:, j]
        mae  = mean_absolute_error(y_true, y_pred)
        rmse = mean_squared_error(y_true, y_pred)
        maes.append(mae); rmses.append(rmse)
        print(f"[{tname}] MAE={mae:.4f} RMSE={rmse:.4f}")

        plot_series(test_dates, y_true, y_pred, f"Real vs Predicho - {tname}")
        plot_scatter(y_true, y_pred, f"Scatter Real vs Predicho - {tname}")
        residuals = (y_pred - y_true)
        plot_residuals(test_dates, residuals, f"Residuales - {tname}")

    print(f"PROMEDIO -> MAE={np.mean(maes):.4f} RMSE={np.mean(rmses):.4f}")

else:
    print("\n=== MODO SINGLE-TARGET (uno por cada objetivo) ===")
    results = []
    for tcol in ALL_TARGETS:
        print(f"\n--- Entrenando target: {tcol} ---")
        y_scaler = MinMaxScaler()
        y_train = y_scaler.fit_transform(train_df[[tcol]].values)
        y_test  = y_scaler.transform(test_df[[tcol]].values)

        Xtr, ytr = make_sequences(X_train_all, y_train, LOOKBACK)
        Xte, yte = make_sequences(X_test_all,  y_test,  LOOKBACK)
        print("Shapes -> Xtr:", Xtr.shape, "ytr:", ytr.shape, "| Xte:", Xte.shape, "yte:", yte.shape)

        model, yhat_scaled, loss_hist = train_eval_model(Xtr, ytr, Xte, yte, out_dim=1)
        yhat_inv = y_scaler.inverse_transform(yhat_scaled)
        yte_inv  = y_scaler.inverse_transform(yte)

        mae  = mean_absolute_error(yte_inv, yhat_inv)
        rmse = mean_squared_error(yte_inv, yhat_inv)
        print(f"[{tcol}] MAE={mae:.4f} RMSE={rmse:.4f}")
        results.append((tcol, float(mae), float(rmse)))

        plot_loss_curve(loss_hist, f"Pérdida (Train MSE) - {tcol}")
        y_true_1d = yte_inv.reshape(-1)
        y_pred_1d = yhat_inv.reshape(-1)
        plot_series(test_dates, y_true_1d, y_pred_1d, f"Real vs Predicho - {tcol}")
        plot_scatter(y_true_1d, y_pred_1d, f"Scatter Real vs Predicho - {tcol}")
        residuals = (y_pred_1d - y_true_1d)
        plot_residuals(test_dates, residuals, f"Residuales - {tcol}")

    print("\n=== RESUMEN SINGLE-TARGET ===")
    for name, mae, rmse in results:
        print(f"{name:45s}  MAE={mae:.4f}  RMSE={rmse:.4f}")
````

````python

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
warnings.filterwarnings("ignore")


TRAIN_PATH = "./sample_data/trainlimpio.csv"
TEST_PATH  = "./sample_data/testlimpio.csv"
DATE_COL   = "date_id"
LOOKBACK   = 64
BATCH_SIZE = 64
EPOCHS     = 60
LR         = 1e-3
DEVICE     = "cuda" if torch.cuda.is_available() else "cpu"
SEED       = 123

np.random.seed(SEED)
torch.manual_seed(SEED)

instruments = {
    'JPX_Gold_Standard_Futures': ['Close', 'High', 'Low', 'Open'],
    'JPX_Gold_Mini_Futures': ['settlement_price', 'High', 'Low', 'Close', 'Open'],
}


def build_targets(instruments_dict):
    cols = []
    for base, fields in instruments_dict.items():
        for f in fields:
            cols.append(f"{base}_{f}")
    seen = set(); ordered = []
    for c in cols:
        if c not in seen:
            seen.add(c); ordered.append(c)
    return ordered

def make_sequences(X, y, lookback):
    Xs, ys = [], []
    for i in range(len(X) - lookback):
        Xs.append(X[i:i+lookback])
        ys.append(y[i+lookback])
    return np.array(Xs, dtype=np.float32), np.array(ys, dtype=np.float32)


train_df = pd.read_csv(TRAIN_PATH).sort_values(DATE_COL).reset_index(drop=True)
test_df  = pd.read_csv(TEST_PATH).sort_values(DATE_COL).reset_index(drop=True)

extra_cols = [c for c in test_df.columns if c not in train_df.columns]
if extra_cols:
    print("Eliminando columnas extra en test:", extra_cols)
    test_df = test_df.drop(columns=extra_cols, errors="ignore")

common_cols = [c for c in train_df.columns if c in test_df.columns]
train_df, test_df = train_df[common_cols], test_df[common_cols]
train_df = train_df.fillna(method="ffill").fillna(method="bfill")
test_df  = test_df.fillna(method="ffill").fillna(method="bfill")

ALL_TARGETS = [t for t in build_targets(instruments) if t in train_df.columns and t in test_df.columns]
print(f"Targets detectados ({len(ALL_TARGETS)}):", ALL_TARGETS)


class TemporalFusionTransformer(nn.Module):
    def __init__(self, input_size, hidden_size=128, num_heads=4, dropout=0.1):
        super().__init__()
        self.input_proj = nn.Linear(input_size, hidden_size)
        self.lstm_enc = nn.LSTM(hidden_size, hidden_size, batch_first=True)
        self.lstm_dec = nn.LSTM(hidden_size, hidden_size, batch_first=True)
        self.attn = nn.MultiheadAttention(hidden_size, num_heads, dropout=dropout, batch_first=True)
        self.gate = nn.Sequential(nn.Linear(hidden_size, hidden_size), nn.Sigmoid())
        self.fc = nn.Linear(hidden_size, 1)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        x = self.input_proj(x)
        enc_out, _ = self.lstm_enc(x)
        dec_out, _ = self.lstm_dec(enc_out)
        attn_out, _ = self.attn(dec_out, enc_out, enc_out)
        gated = self.gate(attn_out) * attn_out + (1 - self.gate(attn_out)) * dec_out
        out = self.fc(self.dropout(gated[:, -1, :]))
        return out


results = []

for target in ALL_TARGETS:
    print(f"\n============= ENTRENANDO TARGET: {target} =============")

    num_cols = [c for c in train_df.columns if np.issubdtype(train_df[c].dtype, np.number) and c != DATE_COL]
    X_train_df, y_train_df = train_df[num_cols], train_df[[target]]
    X_test_df,  y_test_df  = test_df[num_cols],  test_df[[target]]

    x_scaler, y_scaler = MinMaxScaler(), MinMaxScaler()
    X_train_scaled = x_scaler.fit_transform(X_train_df)
    X_test_scaled  = x_scaler.transform(X_test_df)
    y_train_scaled = y_scaler.fit_transform(y_train_df)
    y_test_scaled  = y_scaler.transform(y_test_df)

    Xtr, ytr = make_sequences(X_train_scaled, y_train_scaled, LOOKBACK)
    Xte, yte = make_sequences(X_test_scaled, y_test_scaled, LOOKBACK)

    Xtr_t = torch.tensor(Xtr, dtype=torch.float32)
    ytr_t = torch.tensor(ytr, dtype=torch.float32)
    Xte_t = torch.tensor(Xte, dtype=torch.float32)
    yte_t = torch.tensor(yte, dtype=torch.float32)
    loader = DataLoader(TensorDataset(Xtr_t, ytr_t), batch_size=BATCH_SIZE, shuffle=True)

    model = TemporalFusionTransformer(input_size=Xtr.shape[-1]).to(DEVICE)
    opt = optim.Adam(model.parameters(), lr=LR)
    crit = nn.MSELoss()

    best_loss = float("inf")
    for epoch in range(1, EPOCHS + 1):
        model.train()
        total_loss = 0
        for xb, yb in loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            opt.zero_grad()
            pred = model(xb)
            loss = crit(pred, yb)
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            opt.step()
            total_loss += loss.item() * xb.size(0)
        total_loss /= len(loader.dataset)
        if epoch % 5 == 0 or epoch == 1:
            print(f"Época {epoch:03d}/{EPOCHS}  - Train MSE: {total_loss:.6f}")

    model.eval()
    with torch.no_grad():
        preds = []
        for i in range(0, len(Xte_t), 512):
            xb = Xte_t[i:i+512].to(DEVICE)
            preds.append(model(xb).cpu().numpy())
        yhat = np.vstack(preds)

    yhat_inv = y_scaler.inverse_transform(yhat)
    yte_inv  = y_scaler.inverse_transform(yte)

    mse = mean_squared_error(yte_inv, yhat_inv)
    mae = mean_absolute_error(yte_inv, yhat_inv)
    results.append((target, mse, mae))
    print(f"[{target}] MSE={mse:.4f}  MAE={mae:.4f}")

    plt.figure(figsize=(11,4))
    plt.plot(yte_inv, label="Real")
    plt.plot(yhat_inv, label="Predicho")
    plt.title(f"TFT Puro (PyTorch) - {target}")
    plt.legend()
    plt.tight_layout()
    plt.show()


print("\n=== RESUMEN FINAL ===")
for name, mse, mae in results:
    print(f"{name:45s}  MSE={mse:.4f}  MAE={mae:.4f}")
````
